```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Create a sample DataFrame
df = pd.DataFrame({
    'StudentID': [1, 2, 3, 4, 5, 6],
    'Study_Hours': [7, 8, 5, 6, 9, 10],
    'Marks': [85, 90, 75, 80, 95, 92],
    'Attendance': [90, 95, 80, 85, 50, 88],
    'Gender': ['M', 'F', 'M', 'F', 'M', 'F'],
    'Section': ['A', 'B', 'A', 'D', 'C', np.nan]
})
```

```python
plt.figure(figsize=(5,4))
sns.barplot(data = df, x="Gender", y="Marks")
plt.title("Avg Marks by Gender")
plt.xlabel("Gender")
plt.ylabel("Marks")
plt.show()
```

```python
plt.figure(figsize=(3,4))
sns.pairplot(df)
plt.title("Pairplot for Numerical Columns")
plt.show()
```

```python
plt.figure(figsize=(5,4))
sns.jointplot(data=df, x="Study_Hours", y="Marks", kind='reg')
plt.show()
```

```python
plt.figure(figsize=(5,4))
sns.boxplot(data=df, x="Section", y="Marks")
plt.title("Boxplot of Marks by Section")
plt.xlabel("Section")
plt.ylabel("Marks")
plt.show()
```

```python
df2 = df1.select_dtypes(include=['Int64', 'float64'])
corr = df2.corr()
plt.figure(figsize=(5,4))
sns.heatmap(corr, annot=True, cmap='coolwarm', linewidths=0.5)
plt.title("Correlation Heatmap")
plt.show()
```
